# Получить кадры из видео для формирования данных для обучения.

```python
import os
import cv2
```

```python
labels_input_dir = '../../DataSet/www/Test/labels/'
labels_output_dir = '../../DataSet/www/dullHd/labels/'

os.makedirs(labels_output_dir, exist_ok=True)

scale_x = 1920 / 3840
scale_y = 1080 / 2160

for file in os.listdir(labels_input_dir):
    if file.endswith('.txt'):
        input_path = os.path.join(labels_input_dir, file)
        output_path = os.path.join(labels_output_dir, file)
        
        with open(input_path, 'r') as f:
            lines = f.readlines()
        
        new_lines = []
        for line in lines:
            parts = line.strip().split()
            cls, x_center, y_center, width, height = map(float, parts)
            
            # если координаты были в пикселях:
            x_center *= scale_x
            y_center *= scale_y
            width *= scale_x
            height *= scale_y
            
            # теперь нормализуем
            x_center /= 1920
            y_center /= 1080
            width /= 1920
            height /= 1080
            
            new_line = f"{int(cls)} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}\n"
            new_lines.append(new_line)
        
        with open(output_path, 'w') as f:
            f.writelines(new_lines)
```

```python
input_dir =  '../../DataSet/www/1/images/'
output_dir =  '../../DataSet/www/2/images/'
os.makedirs(output_dir, exist_ok=True)

for filename in os.listdir(input_dir):
    if filename.endswith('.jpg') or filename.endswith('.png') or filename.endswith('.JPG'):
        path = os.path.join(input_dir, filename)
        img = cv2.imread(path)

        # Ресайз до FullHD
        resized = cv2.resize(img, (1920, 1080))

        # Получаем имя файла без расширения
        name, _ = os.path.splitext(filename)

        # Формируем путь с новым расширением .jpg
        output_path = os.path.join(output_dir, f'{name}.jpg')

        # Сохранение
        cv2.imwrite(output_path, resized)
```

```python
import os
import cv2
import numpy as np

input_dir = '../../DataSet/www/2/images/'
output_dir = '../../DataSet/www/2/images/'

os.makedirs(output_dir, exist_ok=True)

for filename in os.listdir(input_dir):
    if filename.lower().endswith(('.jpg', '.png')):
        input_path = os.path.join(input_dir, filename)
        img = cv2.imread(input_path)

        if img is None:
            print(f'Не удалось прочитать: {filename}')
            continue

        # Нормализация: 0–1
        img_normalized = img.astype(np.float32) / 255.0

        # (Здесь можно вставить доп. обработку, если нужно)

        # Обратно в uint8, чтобы сохранить
        img_uint8 = (img_normalized * 255).astype(np.uint8)

        # Сохраняем как .jpg независимо от исходного формата
        base_name = os.path.splitext(filename)[0]
        output_path = os.path.join(output_dir, f"{base_name}.jpg")
        cv2.imwrite(output_path, img_uint8)

print("Готово! Все изображения сохранены в JPG и нормализованы.")
```
